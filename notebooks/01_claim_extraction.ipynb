# 01 — Claim Extraction

Learn how to split an LLM answer into individual claims.

```python
import spacy
```

## 1. Import spaCy

spaCy will help us process the text.

```python
nlp = spacy.load("en_core_web_sm")
```

## 2. Load the NLP Model

Load the English language model.

```python
text = """
The Earth is the third planet from the Sun.
It has one natural satellite called the Moon.
The Moon orbits the Earth.
"""
```

## 3. Create a Sample Text

Use a few simple sentences to test sentence segmentation.

```python
doc = nlp(text)
```

## 4. Split Text into Sentences

spaCy identifies individual sentences for us.

```python
def extract_claims(text):
    doc = nlp(text)
    claims = []
    for sentence in doc.sents:
        claims.append(sentence.text.strip())
    return claims
```

## 5. Create a Claim Extraction Function

For this first version, one sentence = one claim.

```python
answer = """
The James Webb Space Telescope was launched in 2021.
It was launched using an Ariane 5 rocket.
The telescope is approximately 1.5 million kilometers from Earth.
The telescope was launched from India.
"""
claims = extract_claims(answer)
for i, claim in enumerate(claims, start=1):
    print(f"Claim {i}: {claim}")
```

## 7. Observation

The baseline works for simple answers, but one sentence can contain more than one claim.
